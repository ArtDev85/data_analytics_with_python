### Task 1

```python
df = pd.DataFrame({
    "student": ["Anna", "Ben", "Chris", "Diana", "Eva"],
    "n_dependants": [4, 5, 7, 1, 3],
    "Age": [77, 55, 81, 64, 93]
})
```

```python
def family_size_group(n_dependants):
    if n_dependants == 0:
        return "No dependants"
    elif n_dependants <= 3:
        return "Small family"
    else:
        return "Large family"
```

```python
df["family_size_group"] = df["n_dependants"].apply(family_size_group)
df
```

```python
def senior_flag(Age):
    if Age >= 60:
        return "Senior"
    else:
        return "Not senior"
```

```python
df.head()
```

```python
df["senior_flag"] = df["Age"].apply(senior_flag)
df
```
